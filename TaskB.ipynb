```python
import boto3
import pandas as pd
import io
```

```python
s3_client = boto3.client("s3")
```

```python
bucket = "practice-datalake-carlos-jaramillo"
prefix = "raw-zone/weather/"

response = s3_client.list_objects_v2(
            Bucket=bucket,
            Prefix=prefix
        )
```

```python
if "Contents" in response:
    consolidate_weather = pd.DataFrame()
    
    for file in response["Contents"]:
        print(file["Key"])
        obj = s3_client.get_object(Bucket=bucket, Key=file["Key"])
        df = pd.read_csv(io.BytesIO(obj["Body"].read()))
        consolidate_weather = pd.concat([consolidate_weather, df], axis=0, ignore_index=True)
else:
    print("No files found in raw-zone/weather/")
```

```python
consolidate_weather
```

```python
number_readings_collected = len(consolidate_weather)
print("Número de registros encontrados: ", number_readings_collected)
```

```python
mean_temperature = consolidate_weather["temperature_C"].mean()
print("Temperatura promedio: ", mean_temperature)
```

```python
min_humidity = consolidate_weather["humidity_pct"].min()
max_humidity = consolidate_weather["humidity_pct"].max()
print("Humedad mínima: ", min_humidity)
print("Humedad máxima: ", max_humidity)
```
