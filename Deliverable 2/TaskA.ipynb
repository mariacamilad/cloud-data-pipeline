```python
import boto3
```

```python
s3_client = boto3.client("s3")
```

```python
bucket = "practice-datalake-carlos-jaramillo"
prefix = "raw-zone/weather/"

response = s3_client.list_objects_v2(
            Bucket=bucket,
            Prefix=prefix
        )
```

```python
if "Contents" in response:
    print("Files found:")
    for obj in response["Contents"]:
        print(obj["Key"])
else:
    print("No files found in raw-zone/weather/")
```
